# Persiapan Data: Pembersihan, Label AQI, dan Fitur (PySpark + HDFS)
**Kelompok 07 | Mahadata**

Input : `hdfs://namenode:9000/openaq/raw/GabunganDataset_BigData.csv` (mentah, format long)
Output: Parquet `hdfs://namenode:9000/openaq/clean/aqi_fitur` (satu baris per lokasi per jam, berisi label AQI dan fitur)

Jalankan sel dari atas ke bawah. Komentar "Nilai acuan" di tiap sel dipakai untuk memeriksa hasil.

## 0. Setup

```python
import numpy as np
from pyspark.sql import SparkSession, functions as F, Window

spark = (SparkSession.builder
         .appName("AQI_PersiapanData")
         .config("spark.sql.session.timeZone", "UTC")
         .getOrCreate())
spark.sparkContext.setLogLevel("WARN")

RAW   = "hdfs://namenode:9000/openaq/raw/GabunganDataset_BigData.csv"
EDA   = "hdfs://namenode:9000/openaq/raw/GabunganDataset_BigData_eda.csv"   # hanya untuk pembanding
CLEAN = "hdfs://namenode:9000/openaq/clean/aqi_fitur"

PARAMS = ["pm1", "pm25", "um003", "temperature", "relativehumidity"]
print("Spark", spark.version)
```

## 1. Baca data mentah dari HDFS

```python
raw = spark.read.csv(RAW, header=True, inferSchema=True)
raw.printSchema()
print("jumlah baris:", raw.count())      # Nilai acuan: 71655
print("jumlah kolom:", len(raw.columns)) # Nilai acuan: 19
raw.show(5, truncate=False)
```

```python
# Nama parameter yang ada di data (harus cocok dengan PARAMS di atas)
raw.groupBy("parameter").count().show()
raw.groupBy("location_name").count().show()
```

## 2. Pilih kolom, ubah waktu, cek duplikasi
Kolom konstan atau redundan (units, min, max, timezone, owner_name, provider, coverage_percent) tidak dipakai,
sesuai keputusan EDA Minggu 9. Yang dipertahankan hanya lokasi, parameter, nilai, dan waktu.

```python
long_df = (raw
           .select("location_name", "parameter", "value", "datetime_from_utc")
           .withColumn("ts", F.to_timestamp("datetime_from_utc"))
           .drop("datetime_from_utc"))

dup = (long_df.groupBy("location_name", "parameter", "ts").count()
       .filter(F.col("count") > 1).count())
print("kunci duplikat (lokasi, parameter, jam):", dup)   # Nilai acuan: 0

# Nilai negatif (tidak wajar untuk sensor ini)
print("nilai negatif:", long_df.filter(F.col("value") < 0).count())
```

## 3. Pivot: long menjadi wide (satu baris per lokasi per jam)

```python
wide = (long_df
        .groupBy("location_name", "ts")
        .pivot("parameter", PARAMS)
        .agg(F.first("value")))

print("baris setelah pivot:", wide.count())   # Nilai acuan: 14331
wide.show(5)
```

## 4. Lengkapi jam yang hilang
Jam yang tidak ada di data ditambahkan sebagai baris kosong (nilai null, tidak diisi nilai karangan),
supaya lag 1 baris benar-benar berarti 1 jam sebelumnya.

```python
rentang = wide.groupBy("location_name").agg(F.min("ts").alias("awal"), F.max("ts").alias("akhir"))
grid = rentang.select(
    "location_name",
    F.explode(F.sequence("awal", "akhir", F.expr("INTERVAL 1 HOUR"))).alias("ts"))

data = grid.join(wide, on=["location_name", "ts"], how="left")
print("baris setelah grid jam lengkap:", data.count())   # Nilai acuan: 14406 (7203 per lokasi)
data.groupBy("location_name").count().show()
```

## 5. Bersihkan PM2.5
PM2.5 bernilai 0 dianggap error sensor (alasan di EDA Minggu 9) dan diubah menjadi null.

```python
print("PM2.5 == 0 sebelum:", data.filter(F.col("pm25") == 0).count())   # Nilai acuan: 688

data = data.withColumn("pm25",
                       F.when(F.col("pm25") == 0, F.lit(None).cast("double")).otherwise(F.col("pm25")))

print("PM2.5 null sesudah:", data.filter(F.col("pm25").isNull()).count())   # Nilai acuan: 763 (688 nol + 75 jam hilang)
```

## 6. Validasi terhadap hasil EDA Minggu 9

```python
eda = (spark.read.csv(EDA, header=True, inferSchema=True)
       .withColumn("ts", F.to_timestamp("datetime_from_utc")))

gabung = data.alias("a").join(eda.alias("b"), on=["location_name", "ts"], how="inner")
print("baris cocok:", gabung.count())   # Nilai acuan: 14406
for c in PARAMS:
    beda = gabung.filter(~F.col(f"a.{c}").eqNullSafe(F.col(f"b.{c}"))).count()
    print(f"{c}: selisih {beda}")       # Nilai acuan: 0 untuk semua
```

## 7. Label AQI (PM2.5 jam t, EPA versi terbaru)
0 Baik (<=9,0) | 1 Sedang (<=35,4) | 2 Tidak sehat bagi kelompok sensitif (<=55,4) | 3 Tidak sehat (<=125,4) | 4 Sangat tidak sehat (<=225,4) | 5 Berbahaya

```python
data = data.withColumn(
    "label",
    F.when(F.col("pm25").isNull(), F.lit(None).cast("double"))
     .when(F.col("pm25") <= 9.0, 0.0)
     .when(F.col("pm25") <= 35.4, 1.0)
     .when(F.col("pm25") <= 55.4, 2.0)
     .when(F.col("pm25") <= 125.4, 3.0)
     .when(F.col("pm25") <= 225.4, 4.0)
     .otherwise(5.0))

# Cek awal distribusi kelas per lokasi
data.groupBy("location_name", "label").count().orderBy("location_name", "label").show(20)
# Nilai acuan TTDI: kelas 4 = 172 jam | Tambulaung: kelas 4 = 39 jam | kelas 5 tidak muncul
```

## 8. Fitur
PM2.5 jam yang sama **tidak** dipakai sebagai fitur (mencegah data leakage). Semua pengukuran sensor memakai nilai jam sebelumnya.

```python
w = Window.partitionBy("location_name").orderBy("ts")

for k in (1, 2, 3):
    data = data.withColumn(f"pm25_lag{k}", F.lag("pm25", k).over(w))
data = data.withColumn("pm25_ma3", F.avg("pm25").over(w.rowsBetween(-3, -1)))

for c in ["pm1", "um003", "temperature", "relativehumidity"]:
    data = data.withColumn(f"{c}_lag1", F.lag(c, 1).over(w))

data = (data
        .withColumn("ts_local", F.expr("ts + INTERVAL 8 HOURS"))      # Malaysia UTC+8
        .withColumn("jam", F.hour("ts_local"))
        .withColumn("jam_sin", F.sin(2 * np.pi * F.col("jam") / 24))
        .withColumn("jam_cos", F.cos(2 * np.pi * F.col("jam") / 24))
        .withColumn("is_ttdi", (F.col("location_name") == "Taman Tun Dr. Ismail").cast("double")))

FITUR = ["pm25_lag1", "pm25_lag2", "pm25_lag3", "pm25_ma3",
         "pm1_lag1", "um003_lag1", "temperature_lag1", "relativehumidity_lag1",
         "jam_sin", "jam_cos", "is_ttdi"]

final = (data
         .dropna(subset=FITUR + ["label"])
         .select("location_name", "ts", "ts_local", "pm25", "label", *FITUR))

print("baris akhir (siap model):", final.count())
final.groupBy("location_name").count().show()
final.show(5)
```

## 9. Simpan ke HDFS sebagai Parquet

```python
final.write.mode("overwrite").parquet(CLEAN)

cek = spark.read.parquet(CLEAN)
print("baca ulang dari Parquet:", cek.count(), "baris,", len(cek.columns), "kolom")
cek.printSchema()
```

## 10. Cek akhir
Distribusi kelas per lokasi dan per bagian (latih Jan-Agu, uji Sep-Okt).

```python
batas = F.lit("2026-09-01 00:00:00").cast("timestamp")
(cek.withColumn("bagian", F.when(F.col("ts_local") < batas, "latih").otherwise("uji"))
    .groupBy("location_name", "bagian", "label").count()
    .orderBy("location_name", "bagian", "label")
    .show(40))
```
